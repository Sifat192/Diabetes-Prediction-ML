# Diabetes Prediction using Logistic Regression

## Introduction

Diabetes is a common chronic disease that affects how the body processes blood sugar. Early detection can help reduce serious health complications.

In this notebook, we build a **Logistic Regression model** to predict whether a patient has diabetes using health-related features such as **age, BMI, blood glucose level, HbA1c level, and smoking history**. The project demonstrates a basic **machine learning workflow**, including data preprocessing, feature encoding, model training, and evaluation.

## Dataset

The dataset contains medical and demographic information about patients. The target variable **`diabetes`** indicates whether a patient has diabetes.

**Key features:**

* Gender
* Age
* Hypertension
* Heart Disease
* Smoking History
* BMI
* HbA1c Level
* Blood Glucose Level

The goal is to build a model that **predicts diabetes based on these features**.


## Importing  Libraries

```python
import pandas as pd 
from sklearn import linear_model
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score,confusion_matrix
```

## Loading Dataset

```python
df = pd.read_csv('diabetes_prediction_dataset.csv')
```

## Exploratory Data Analysis

```python
df.head()
```

```python
df.tail()
```

```python
df['gender'] = df['gender'].map({'Male':1,'Female':0})
```

```python
df['smoking_history'] = df['smoking_history'].map({
    'never':0,
    'former':1,
    'current':2,
    'not current':3,
    'ever':4,
    'No Info':5
})
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df['gender']=df['gender'].fillna(df['gender'].mean())
```

```python
df['diabetes'].value_counts()
```

## Model Training

```python
x = df[['gender','age','hypertension','heart_disease','bmi','HbA1c_level','blood_glucose_level','smoking_history']]
```

```python
y = df.diabetes
```

```python
scl = StandardScaler()
x = scl.fit_transform(x)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2)
```

```python
model = linear_model.LogisticRegression()
```

```python
model.fit(x_train,y_train)
```

## Model Prediction

```python
y_pred = model.predict(x_test)
```

```python
model.coef_
```

## Model Evaluation

```python
model.score(x_test,y_test)
```

```python
f1_score(y_test, y_pred)
```

```python
confusion_matrix(y_test,y_pred)
```
